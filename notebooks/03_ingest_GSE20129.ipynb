```python
import GEOparse, pandas as pd

g = GEOparse.get_GEO(geo="GSE20129", destdir="../data/bronze", silent=True)
ph = g.phenotype_data
print(pd.crosstab(ph['platform_id'], ph['characteristics_ch1.0.disease state']))

for k, v in g.gpls.items():
    print("\n", k, v.table.shape, v.table.columns.tolist()[:12])

for pid in ph['platform_id'].unique():
    s = ph.index[ph['platform_id'] == pid][0]
    t = g.gsms[s].table
    print("\n", pid, s, t.columns.tolist(), t['VALUE'].describe().round(2).to_dict())
```
